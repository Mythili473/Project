```python
import sqlite3
from datetime import date

# Connect to SQLite DB
conn = sqlite3.connect("mood_journal.db")
cursor = conn.cursor()

# Create table
cursor.execute('''
CREATE TABLE IF NOT EXISTS journal (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    entry_date TEXT NOT NULL,
    mood TEXT NOT NULL,
    note TEXT,
    weather TEXT
)
''')
conn.commit()

# Add sample entries if table is empty
def seed_sample_data():
    cursor.execute("SELECT COUNT(*) FROM journal")
    count = cursor.fetchone()[0]
    if count == 0:
        sample_entries = [
            ("2025-07-18", "Happy", "Had a great day at work, felt productive.", "Sunny"),
            ("2025-07-19", "Tired", "Stayed up late watching movies. Need rest.", "Cloudy"),
            ("2025-07-20", "Anxious", "Upcoming deadline is stressing me out.", "Rainy"),
            ("2025-07-21", "Relaxed", "Did yoga in the morning and felt good all day.", "Clear"),
            ("2025-07-22", "Sad", "Missed a call from a close friend.", "Windy")
        ]
        cursor.executemany('''
            INSERT INTO journal (entry_date, mood, note, weather)
            VALUES (?, ?, ?, ?)
        ''', sample_entries)
        conn.commit()
        print("📥 Sample journal entries added.\n")

# Add new journal entry
def add_entry():
    print("\n➕ Add Journal Entry")
    entry_date = input("Enter date (YYYY-MM-DD) [Leave blank for today]: ") or str(date.today())
    mood = input("Mood (e.g., Happy, Sad): ")
    note = input("Note (optional): ")
    weather = input("Weather (e.g., Rainy, Sunny): ")

    cursor.execute('''
        INSERT INTO journal (entry_date, mood, note, weather)
        VALUES (?, ?, ?, ?)
    ''', (entry_date, mood, note, weather))
    conn.commit()
    print("✅ Entry saved.\n")

# View entries
def view_entries():
    print("\n📚 View Journal Entries")
    cursor.execute("SELECT * FROM journal ORDER BY entry_date DESC")
    entries = cursor.fetchall()

    if not entries:
        print("No entries found.")
        return

    for entry in entries:
        print(f"\n📅 Date     : {entry[1]}")
        print(f"🙂 Mood     : {entry[2]}")
        print(f"📝 Note     : {entry[3]}")
        print(f"☁ Weather  : {entry[4]}")
        print("-" * 40)

# Delete entry
def delete_entry():
    print("\n🗑 Delete Entry")
    entry_date = input("Enter the date to delete (YYYY-MM-DD): ")
    cursor.execute("SELECT * FROM journal WHERE entry_date = ?", (entry_date,))
    result = cursor.fetchone()
    if result:
        confirm = input("Are you sure you want to delete this entry? (y/n): ").lower()
        if confirm == "y":
            cursor.execute("DELETE FROM journal WHERE entry_date = ?", (entry_date,))
            conn.commit()
            print("✅ Entry deleted.\n")
    else:
        print("❌ No entry found for that date.\n")

# Update entry
def update_entry():
    print("\n✏ Update Entry")
    entry_date = input("Enter the date to update (YYYY-MM-DD): ")
    cursor.execute("SELECT * FROM journal WHERE entry_date = ?", (entry_date,))
    entry = cursor.fetchone()

    if entry:
        new_mood = input(f"New mood (leave blank to keep '{entry[2]}'): ") or entry[2]
        new_note = input(f"New note (leave blank to keep existing): ") or entry[3]
        cursor.execute('''
            UPDATE journal SET mood = ?, note = ? WHERE entry_date = ?
        ''', (new_mood, new_note, entry_date))
        conn.commit()
        print("✅ Entry updated.\n")
    else:
        print("❌ No entry found for that date.\n")

# Menu
def main():
    seed_sample_data()
    while True:
        print("\n📖 Mood & Journal Logger")
        print("1. Add Entry")
        print("2. View Entries")
        print("3. Update Entry")
        print("4. Delete Entry")
        print("5. Exit")
        choice = input("Choose an option (1-5): ")

        if choice == "1":
            add_entry()
        elif choice == "2":
            view_entries()
        elif choice == "3":
            update_entry()
        elif choice == "4":
            delete_entry()
        elif choice == "5":
            print("👋 Goodbye! Take care of yourself.")
            break
        else:
            print("❗ Invalid option. Try again.")

    conn.close()

if __name__ == "__main__":
    main()
```
